```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')

word="good"

synonyms=set()
antonyms=set()

for syn in wordnet.synsets(word):
    for l in syn.lemmas():
        synonyms.add(l.name())
        if l.antonyms():
            antonyms.add(l.antonyms()[0].name())

print(synonyms)
print(antonyms)
```
